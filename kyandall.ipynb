# Have I Been Flocked Analysis

## Question

How is Flock ALPR technology being used by law enforcement based on
the search reasons recorded in publicly available audit logs?

## Approach

I will use data from Have I Been Flocked and analyze:
- the most common reasons for ALPR searches
- how frequently searches occur
- patterns in search reasons
- missing or unusual search reasons

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("agency-3179-2026-09-09.csv",low_memory=False)

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df["reason"].value_counts().head(20)
```

```python
df["reason_clean"] = (
    df["reason"]
    .fillna("Missing")
    .astype(str)
    .str.strip()
    .str.lower()
)
```

```python
df["reason_clean"].value_counts().head(20)
```

```python
df["reason_clean"].value_counts().head(15).plot(kind="bar")

plt.title("Most Common Knoxville PD Flock Search Reasons")
plt.xlabel("Search Reason")
plt.ylabel("Number of Searches")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df["search_time_utc"] = pd.to_datetime(
    df["search_time_utc"],
    errors="coerce"
)
```

```python
df["month"] = df["search_time_utc"].dt.to_period("M")
```

```python
searches_per_month = df.groupby("month").size()

searches_per_month
```

```python
searches_per_month.plot()

plt.title("Knoxville PD Flock Searches Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Searches")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
